```python
import pandas as pd
```

```python
actas_nombres = ['actas_detalle_1.csv', 'actas_detalle_2.csv']

actas = pd.concat([pd.read_csv(acta, index_col='acta_detalle_id') for acta in actas_nombres])
actas.head()
```

```python
bloques = pd.read_csv('bloques.csv')
bloques['diputado_nombre'] = bloques['diputado_apellido'] + bloques['diputado_nombres']
bloques = bloques.set_index('diputado_nombre', verify_integrity=True)
bloques.head()
```

```python
expedientes_nombres = ['expedientes_1.csv', 'expedientes_2.csv']

expedientes = pd.concat([pd.read_csv(acta, index_col='acta_id') for acta in expedientes_nombres])
expedientes.head()
```

```python
df = actas.join(expedientes, on='acta_id')
df = df[df['titulo'].notna()] # Filtro actas con titulo
df = df[df['diputado_nombre'].notna()] # Filtro detalles con diputado
df
```

```python
import pandas as pd
```

```python
actas_nombres = ['actas_detalle_1.csv', 'actas_detalle_2.csv']

actas = pd.concat([pd.read_csv(acta, index_col='acta_detalle_id') for acta in actas_nombres])
actas.head()
```

```python
def APELLIDOS_nombre(nombres):
    if not isinstance(nombres, list):
        return nombres
    
    res = ''
    for nombre in nombres:
        res +=  ' ' + nombre
        if nombre.upper() != nombre:
            break
    return res.strip()
```

```python
bloques = pd.read_csv('bloques.csv')
bloques['diputado_nombre'] = bloques['diputado_apellido'] + ' ' + bloques['diputado_nombres'].str.strip() # .str.split(' ').apply(lambda r: r[0])
bloques = bloques.set_index('diputado_nombre', verify_integrity=True)
bloques.head()
```

```python
expedientes_nombres = ['expedientes_1.csv', 'expedientes_2.csv']

expedientes = pd.concat([pd.read_csv(acta, index_col='acta_id') for acta in expedientes_nombres])
expedientes.head()
```

```python
df = actas.join(expedientes, on='acta_id')
df = df[df['titulo'].notna()] # Filtro actas con titulo
df = df[df['diputado_nombre'].notna()] # Filtro detalles con diputado
df
```

```python
bloques_nombres = set(bloques.index)
actas_nombres = set(actas['diputado_nombre'])
```

```python
df = df.join(bloques, on='diputado_nombre')
df.head()
```

```python
interbloques = pd.read_csv('interbloques.csv')
interbloques['diputado_nombre'] = interbloques['diputado_apellido'] + ' ' + interbloques['diputado_nombre'].str.strip() # .str.split(' ').apply(lambda r: r[0])

interbloques = interbloques.set_index('diputado_nombre', verify_integrity=True)
interbloques
```

```python
df = df.join(interbloques['interbloque'], on='diputado_nombre')
df.head()
```
